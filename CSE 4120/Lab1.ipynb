```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris= load_iris()
print(iris)
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print(df)
```

```python
df['target'] = iris.target
df['species'] = df['target'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
print(df['target'])
```

Task 1. Check for missing values in the Iris dataset
Fill missing values with the mean

```python
df.loc[80:100, 'sepal length (cm)'] = np.nan
print(df.isnull().sum())
median_value = df['sepal length (cm)'].median()
df['sepal length (cm)'] = df['sepal length (cm)'].fillna(median_value)
print("\nAfter filling missing values:")
print(df.isnull().sum())
```

Task 2 .Plot histogram for “petal length (cm)"

```python
print(df['sepal length (cm)'])
plt.hist(df['sepal length (cm)'], bins=15, color='blue')
plt.title("Histogram of Sepal Length")
plt.show()
```

Task 3. Plot boxplots of “sepal width (cm)” for each species

```python
sns.boxplot(x='species', y='sepal length (cm)', data=df)
plt.show()
```
