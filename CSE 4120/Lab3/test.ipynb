```python
import tensorflow as tf
import torch

# Test TensorFlow GPU
tf_gpu_available = tf.config.list_physical_devices('GPU')
print("TensorFlow GPU available:", bool(tf_gpu_available))
if tf_gpu_available:
    gpu = tf_gpu_available[0]
    print("TensorFlow GPU name:", gpu.name)
    print("TensorFlow GPU device type:", gpu.device_type)
    print("Total number of TensorFlow GPUs:", len(tf_gpu_available))

# Test PyTorch GPU
torch_gpu_available = torch.cuda.is_available()
print("PyTorch GPU available:", torch_gpu_available)
if torch_gpu_available:
    print("PyTorch GPU name:", torch.cuda.get_device_name(0))
```
