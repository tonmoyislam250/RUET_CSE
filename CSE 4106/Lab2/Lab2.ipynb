Implement Image Negative, Log transformation, Gamma Correction Thresholding Contrast Strechting

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('lena_gray.jpg', 0)
```

1. Image Negative

```python
img_negative_manual = np.zeros_like(img)
rows, cols = img.shape
for i in range(rows):
    for j in range(cols):
        img_negative_manual[i, j] = 255 - img[i, j]

plt.imshow(img_negative_manual, cmap='gray')
plt.title('Image Negative')
plt.axis('off')
plt.show()
```

2. Log Transformation

```python
img_log_manual = np.zeros_like(img, dtype=np.float32)
c = 255 / np.log(1 + np.max(img))
for i in range(rows):
    for j in range(cols):
        img_log_manual[i, j] = c * np.log(1 + img[i, j])

img_log_manual = np.array(img_log_manual, dtype=np.uint8)

plt.imshow(img_log_manual, cmap='gray')
plt.title('Log Transformation')
plt.axis('off')
plt.show()
```

3. Gamma Correction

```python
gamma = 0.4
img_gamma_manual = np.zeros_like(img, dtype=np.float32)

for i in range(rows):
    for j in range(cols):
        img_gamma_manual[i, j] = 1 * (img[i, j]** gamma)
        # print(f"Pixel ({i}, {j}) value before gamma correction: {img[i, j]}")
        # print(f"Pixel ({i}, {j}) value after gamma correction: {img_gamma_manual[i, j]}")

img_gamma_manual = np.array(img_gamma_manual, dtype=np.uint8)

# print("Original Image (img):")
# for row in img:
#     print(' '.join(f"{val:3d}" for val in row))

# print("\nImage Negative (img_negative_manual):")
# for row in img_negative_manual:
#     print(' '.join(f"{val:3d}" for val in row))

plt.imshow(img_gamma_manual, cmap='gray')
plt.title('Gamma Correction (Manual)')
plt.axis('off')
plt.show()
```

4. Thresholding

```python
threshold_value = 220
img_threshold_manual = np.zeros_like(img, dtype=np.uint8)

for i in range(rows):
    for j in range(cols):
        if img[i, j] > threshold_value:
            img_threshold_manual[i, j] = 255
        else:
            img_threshold_manual[i, j] = 0

plt.imshow(img_threshold_manual, cmap='gray')
plt.title('Thresholding')
plt.axis('off')
plt.show()
```

5. Contrast Stretching

```python
img_contrast_stretch_manual = np.zeros_like(img, dtype=np.float32)
min_val = np.min(img)
max_val = np.max(img)
print(f"Min: {min_val}, Max: {max_val}")
for i in range(rows):
    for j in range(cols):
        img_contrast_stretch_manual[i, j] = (img[i, j] - min_val) / (max_val - min_val) * 255

img_contrast_stretch_manual = np.array(img_contrast_stretch_manual, dtype=np.uint8)

plt.imshow(img_contrast_stretch_manual, cmap='gray')
plt.title('Contrast Stretching')
plt.axis('off')
plt.show()
```
