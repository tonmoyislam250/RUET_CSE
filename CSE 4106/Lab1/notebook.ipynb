```python
image_bgr = cv2.imread('images/image1.jpg')
plt.subplot(1, 2, 1)
plt.imshow(image_bgr)
```

```python
plt.subplot(1, 2, 1)
plt.title('Image in Matplotlib')
plt.imshow(image_bgr)

image_hsv = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2HSV)

cv2.imshow('HSV Image', image_hsv)
cv2.imwrite('output_image_hsv.jpg', image_hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()

plt.subplot(1, 2, 2)
plt.title('HSV Image in Matplotlib')
plt.imshow(image_hsv)

plt.savefig('images\\matplotlib_plots.png')  # Save the figure as an image
plt.show()
```
